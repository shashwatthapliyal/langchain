```python
import os
from google.colab import userdata
```

```python
os.environ["GOOGLE_API_KEY"]=userdata.get("GEMINI_API")
```

```python
cricketers = [
    "Virat Kohli is an Indian batsman known for his aggressive batting.",
    "Rohit Sharma is an Indian cricketer famous for his powerful pull shots.",
    "MS Dhoni is a former Indian captain known for his wicketkeeping and finishing skills.",
    "Sachin Tendulkar is a legendary Indian batsman known as the Master Blaster.",
    "Jasprit Bumrah is an Indian fast bowler famous for his unique bowling action.",
    "Ben Stokes is an English all-rounder known for his match-winning performances.",
    "Hardik Pandya is an Indian all-rounder known for his batting and fast bowling.",
    "Kane Williamson is a New Zealand batsman known for his calm and consistent batting.",
    "Babar Azam is a Pakistani batsman known for his elegant batting technique.",
    "Pat Cummins is an Australian fast bowler and former captain."
]
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
embeddings=GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-2-preview"
)
```

```python
vector=embeddings.embed_documents(cricketers)
```

```python
query="Who is english the all rounder ?"
```

```python
query_vector=embeddings.embed_query(query)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarities=cosine_similarity([query_vector],vector)[0]
```

```python
print(similarities)
```

```python
max=similarities[0]
idx=0
```

```python
for i in range(1,10):
  if(similarities[i]>max):
    idx=i
    max=similarities[i]
```

```python
print(cricketers[idx])
```
